```python
df_clean
```

```python
x = df_clean.drop('depression_label',axis=1)
y= df_clean['depression_label']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test, y_train ,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model_log = LogisticRegression()
model_log.fit(x_train, y_train)
```

```python
y_prev = model_log.predict(x_test)
```

```python
from sklearn.metrics import classification_report,accuracy_score, precision_score
x = classification_report(y_test, y_prev)
acc = accuracy_score(y_test, y_prev)
pre = precision_score(y_test,y_prev)
print("clss",x,"accuracy",acc,"precision",pre)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = RandomForestClassifier(
    n_estimators=500,
    max_depth=10,
    min_samples_split=5,
    random_state=42
)
```

```python
rf_model.fit(x_train, y_train)
```

```python
y_pred = rf_model.predict(x_test)
```

```python
print(accuracy_score(y_test, y_prev))
```

```python
print(precision_score(y_test, y_pred))
```

```python
from imblearn.over_sampling import SMOTE
```

```python
smote = SMOTE(random_state=42,sampling_strategy='minority')
x_train_smote, y_train_smote = smote.fit_resample(x_train, y_train)
```
